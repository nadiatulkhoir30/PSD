## Data Understanding


Pada tahap ini adalah tahap yang kedua dalam CRISP-DM, dimana pada proses ini adalah mengumpulkan, mengeksploari dan mengevaluasi datanya. 

### Membuat Database IRIS di PSQL dan mengambil datanya menggunakan Power BI

1. sebelum membuat database di postgree, pastikan sudah menginstall apk postgree versi terbarunya. Bisa langsung klik link ini untuk mendownload postgree https://www.enterprisedb.com/downloads/postgres-postgresql-downloads

2. klik server, lalu isikan username dan password tadi yang telah diisi saat penginstallan postgree.

3. lalu klik servers, dan klik lagi postgreSQL17, klik lagi databses dan pilih schemas, lalu pilih table dan klik kanan lalu pilih Query tool.

![Deskripsi](images/image.png)

4. copy header dari file csv data iris.

![Deskripsi](images/image-2.png)

5. lalu membuat tabel di psql tadi, dengan memasukkan query seperti berikut.

![Deskripsi](images/image-3.png)

Dimana pada kode itu sudah berhasil dibuat tabel bernama iris2, dimana disitu memasukkan nama header tadi yang sudah di copy dari file csv dan memasukkan tipenya juga. Dimana untuk sepal.length, sepal.width, petal.length, petal.width ini termasuk Tipe Numerik, dan untuk variety dan vnum termasuk tipe kategorikal (memang angka (0,1,2), tapi angka ini cuma kode kategori, bukan nilai numerik yang bermakna untuk dijumlahkan.)

6. klik table lalu klik kanan dan refresh agar table tadi yang sudah dibuat keluar.

![Deskripsi](images/image-4.png)

7. sebelum mengimport data dari file csv, pastikan jika path letak postgre yang diletakkan sudah benar. untuk memastikan jika path nya sudah benar, maka pergi ke menu file dan pilih preferences dan pilih binary path dan lihat di postgre binary path, kita lihat di postgre17 karena tadi yang diinstall versi 17 jadi yang diisi pathnya disitu.

![Deskripsi](images/image-5.png)

8. selanjutnya klik kanan pada table iris2 tadi yang sudah dibuat, lalu pilih dan klik import/export data.

9. lalu di filename tentukan lokasi dari file csv yang digunakan tadi.

![Deskripsi](images/image-6.png)

10. dan di menu options, bacaan header itu di hidupkan karena tadi pada query psqlnya memasukkan nama nama header dari file csv yang digunakan, jika sudah selesai maka klik OK.

![Deskripsi](images/image-7.png)

11. jika ingin melihat isi data dari file csv tadi, bisa dengan mengklik table iris2 dan klik kanan pilih yang query tool dan masukkan query seperti berikut dan tekan f5 untuk menjalankan query.

![Deskripsi](images/image-8.png)

12. selanjutnya akan mengambil data yang ada di postgre menggunakan power bi, dimana yang pertma adalah buka apk power bi dan pilih get data from another source, lalu pilih yang PostgreSQL Database dan klik Connect.

![Deskripsi](images/image-9.png)

13. Masukkan nama server  dan nama database yang digunakan di postgre, lalu klik ok.

![Deskripsi](images/image-10.png)

14. lalu masukkan nama username dan password tadi yang dibuat saat penginstallan postgre.

![Deskripsi](images/image-11.png)

15. lalu pilih table yang sudah dibuat tadi yang bernama iris2, lalu klik load.

![Deskripsi](images/image-12.png)

16. untuk membuat maksimal dari setiap kategori, maka klik Quick Measure lalu pada dropdown select a calculation to create a measure itu pilih yang Max per category dan yang base value dan category diisi sesuai kategori yang ingin di hitung.

![Deskripsi](images/image-16.png)

untuk perhitungan minimal dan rata-rata caranya sama seperti itu.

17. dan untuk hasil perhitungan dari Minimal, Maksimal Dan Rata-Rata adalah berikut hasil visualisasi dari card.

![Deskripsi](images/image-13.png)
